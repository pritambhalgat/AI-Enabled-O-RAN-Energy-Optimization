# O-RAN Energy Optimization — Version 1 (Baseline)

This notebook is the **baseline version** of the O-RAN energy-efficiency project. It reproduces the original fixed-power versus channel-aware heuristic experiment from the repository's `src/simulate.py`.

## What it includes
- 64 base-station antennas and 8 uplink users
- Rayleigh small-scale channel model
- Fixed equal-power allocation
- Channel-gain-aware heuristic allocation
- Shannon sum-throughput and energy-efficiency calculations

## What it does not include
No supervised ML, DRL, explicit multi-user interference calculation, per-user QoS constraint, or live O-RAN RIC integration. This is the reference baseline against which Version 2 should be evaluated.

## Mathematical model
For each user, the baseline computes an effective gain (g_k) from the channel realization and
$
\mathrm{SINR}_k = \frac{p_k g_k}{\sigma^2_{\mathrm{scaled}}},\quad
R_{\mathrm{sum}}=B\sum_k\log_2(1+\mathrm{SINR}_k).
$
The assumed power model is
$
P_{\mathrm{total}}=\frac{\sum_k p_k}{\eta_{PA}}+P_{\mathrm{circuit}},\qquad
EE=\frac{R_{\mathrm{sum}}}{P_{\mathrm{total}}}.
$
